```python
%%sql
select * from album
join artist on album.artistid = artist.artistid
```

```python
%%sql
select g.Name жанр, count(*) треки
from genre g
join track t on g.genreid = t.genreid
group by жанр
```

```python
%%sql
select c.firstname имя, c.lastname фамилия, count(*) заказы 
from customer c
join invoice i on c.customerid = i.invoiceid
group by имя
```

```python
%%sql
select t.name трек, g.name жанр
from track t
join genre g on t.genreid = g.genreid
where g.name = 'Rock'
group by трек 
order by трек
```
